```python
def evalRPN(tokens):
    """
    :type tokens: List[str]
    :rtype: int
    """
    l       = len(tokens)
    symbols = ["+", "-", "*", "/"]
    stacks  = []
    for i in range(l):
        if tokens[i] not in symbols:
            stacks.append(int(tokens[i]))
        elif tokens[i] == "+":
            # mv = stacks[-2] + stacks[-1]
            mv = stacks.pop()
            mv = int(stacks.pop() + mv)
            stacks.append(mv)
        elif tokens[i] == "-":
            mv = stacks.pop()
            mv = int(stacks.pop() - mv)
            stacks.append(mv)
        elif tokens[i] == "*":
            mv = stacks.pop()
            mv = int(stacks.pop() * mv)
            stacks.append(mv)
        else:
            mv = stacks.pop()
            mv = int(stacks.pop() / mv)
            stacks.append(mv)
    return mv
```

```python
def evalRPN(tokens):
    stacks  = []
    operators = {
        "+" : lambda a, b: a + b,
        "-" : lambda a, b: a - b,
        "*" : lambda a, b: a * b,
        "/" : lambda a, b: int(a / b)
    }
    
    for token in tokens:
        if token in operators:
            right   = stacks.pop()
            left    = stacks.pop()
            result = operators[token](left, right)
            stacks.append(result)
        else:
            stacks.append(int(token))
    return stacks.pop()
```
